```python
# correlation
# ============
# Statistical method used to measure the relationship between two variables
# or columns(eg: age and glucose)
# It helps to understand how a change in one column affect another

# Types
# =====

# positive correlation
# =====================
# if value in one column increases the other column values also incresed
# (Direct relationship >>> value in between (0,1))
# negative correlation
# ====================
# As one column value increase the other column value gets decreased(inverse relationship)
# values in between(0,-1)
# Zero correlation
# =================
# There is no relation between other columns as if the value increase or decrease
```
